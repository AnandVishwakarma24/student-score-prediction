```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
hours = [2,3,5,6,8]
marks = [45,55,75,85,95]

plt.scatter(hours, marks)

plt.title("Hours Studied vs Marks")
plt.xlabel("Hours")
plt.ylabel("Marks")
# plt.plot(hours, marks, marker="o")
plt.scatter(hours, marks, color="green")
plt.show()
```

```python
students = ["A","B","C","D"]
marks = [80,65,95,70]

plt.bar(students, marks)

plt.title("Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")
# plt.plot(students, marks, marker="o")
plt.bar(students, marks, color="orange")
plt.figure(figsize=(8,5))
plt.show()
```

```python
months = ["Jan","Feb","Mar","Apr","May"]
sales = [100,150,180,220,260]

plt.plot(months, sales)

plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.grid(True)
plt.plot(months, sales, marker="o")
# plt.plot(months, sales, linestyle="--")
plt.plot(months, sales, color="red")
plt.figure(figsize=(8,5))

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data =pd.read_csv("student_scores.csv")



# Scatter Plot
plt.figure(figsize=(7,5))
plt.scatter(data["Math"], data["Science"])
plt.title("Math vs Science Marks")
plt.xlabel("Math Marks")
plt.ylabel("Science Marks")
plt.grid(True)
plt.show()

# Bar Chart
plt.figure(figsize=(8,5))
plt.bar(data["Name"], data["Math"])
plt.title("Math Marks of Students")
plt.xlabel("Student Name")
plt.ylabel("Math Marks")
plt.show()

# Line Chart
plt.figure(figsize=(8,5))
plt.plot(data["Name"], data["Attendance"], marker="o")
plt.title("Student Attendance")
plt.xlabel("Student Name")
plt.ylabel("Attendance (%)")
plt.grid(True)
plt.show()
```
