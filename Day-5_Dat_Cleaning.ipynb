```python
data = pd.read_csv("employee.csv")
print(data)

#handling missing values 

#NaN => Not a number

print(f"\n",data.isnull()) #true=> empty or not given value



print(f"\n",data.isnull().sum()) #sum of all null values with their respective columns name

#Removes all rows containing missing values
print(data.dropna())

#for permanent changes in the file
# data.dropna(inplace=True)

#fill missing values
print("\n",data.fillna(0)) #it put the NaN =>0

#Fill with Mean
data["Age"] = data["Age"].fillna(data["Age"].mean())

#Fill with Median
data["Age"] = data["Age"].fillna(data["Age"].median())


#Fill with Mode
data["City"] = data["City"].fillna(data["City"].mode()[0])


#Forward Fill
print(data.ffill())

#Backward Fill
print(data.bfill())


#Remove Duplicates

#Check Duplicate Rows
print(data.duplicated())

#Count Duplicate Rows
print(data.duplicated().sum())

#Remove Duplicate Rows
print(data.drop_duplicates())

#for permanent changes 
# print(data.drop_duplicates(inplace=True))

#Remove Duplicates Using Specific Columns
print(data.drop_duplicates(subset=["Name"]))



```

```python
import pandas as pd

# Load data
df = pd.read_csv("employee.csv")

# Inspect
print(df.head())
print(df.info())
print(df.describe())

# Check missing values
print(df.isnull().sum())

# Fill missing values
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Salary"] = df["Salary"].fillna(df["Salary"].median())
df["City"] = df["City"].fillna(df["City"].mode()[0])

# Remove duplicates
df.drop_duplicates(inplace=True)

# Verify
print(df.isnull().sum())
print(df.duplicated().sum())

# Save cleaned dataset
df.to_csv("employee_cleaned.csv", index=False)
```
